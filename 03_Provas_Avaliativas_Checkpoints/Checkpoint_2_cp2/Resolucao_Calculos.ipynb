**Título:** Cinemática de Manipuladores Robóticos e Parâmetros Denavit-Hartenberg (DH)

> **Enunciado Original:**
> Determinar o modelo cinemático direto de um manipulador robótico industrial planar com 2 graus de liberdade (2R) com elos l1 = 0,5 m e l2 = 0,4 m. Calcular a matriz de transformação homogênea global T_2^0 e a posição cartesiana (x, y) do efetuador para as variáveis de junta θ1 = 30° e θ2 = 45°.

### Formulação Matemática Formal
$$T_i^{i-1} = \begin{bmatrix} \cos\theta_i & -\sin\theta_i\cos\alpha_i & \sin\theta_i\sin\alpha_i & a_i\cos\theta_i \\ \sin\theta_i & \cos\theta_i\cos\alpha_i & -\cos\theta_i\sin\alpha_i & a_i\sin\theta_i \\ 0 & \sin\alpha_i & \cos\alpha_i & d_i \\ 0 & 0 & 0 & 1 \end{bmatrix}$$


### Passo 1: Tabela de Parâmetros de Denavit-Hartenberg
Para o robô planar 2R: Junta 1: \theta_1, d_1 = 0, a_1 = 0,5\text{ m}, \alpha_1 = 0^\circ. Junta 2: \theta_2, d_2 = 0, a_2 = 0,4\text{ m}, \alpha_2 = 0^\circ.


### Passo 2: Matrizes de Transformação Individual
Calculando T_1^0 para a base à junta 1 e T_2^1 da junta 1 ao efetuador com rotação pura em Z e translação a_i.


### Passo 3: Multiplicação das Matrizes Homogêneas
A matriz global é T_2^0 = T_1^0 \cdot T_2^1. As coordenadas do efetuador são: x = a_1 \cos\theta_1 + a_2 \cos(\theta_1 + \theta_2) e y = a_1 \sin\theta_1 + a_2 \sin(\theta_1 + \theta_2).


### Passo 4: Avaliação Numérica dos Ângulos
Substituindo \theta_1 = 30^\circ e \theta_2 = 45^\circ (\theta_1 + \theta_2 = 75^\circ): x = 0,5 \cos(30^\circ) + 0,4 \cos(75^\circ) = 0,5(0,866) + 0,4(0,2588) = 0,433 + 0,1035 = 0,5365\text{ m}.


### Passo 5: Posição do Efetuador e Espaço de Trabalho
Calculando y: y = 0,5 \sin(30^\circ) + 0,4 \sin(75^\circ) = 0,5(0,5) + 0,4(0,9659) = 0,25 + 0,3864 = 0,6364\text{ m}. Efetuador em P = (0,5365; 0,6364)\text{ m}, dentro do raio de alcance R_{max} = 0,9\text{ m}.


```python
import numpy as np
import matplotlib.pyplot as plt

l1, l2 = 0.5, 0.4
theta1_deg, theta2_deg = 30, 45
t1 = np.radians(theta1_deg)
t2 = np.radians(theta2_deg)
x0, y0 = 0, 0
x1, y1 = l1 * np.cos(t1), l1 * np.sin(t1)
x2, y2 = x1 + l2 * np.cos(t1 + t2), y1 + l2 * np.sin(t1 + t2)
# Espaço de trabalho
ang = np.linspace(0, 2*np.pi, 200)
r_max = l1 + l2
r_min = np.abs(l1 - l2)
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#182232')
ax.set_facecolor('#0f141c')
ax.fill(r_max * np.cos(ang), r_max * np.sin(ang), color='#06b6d4', alpha=0.1, label='Workspace Máximo (R=0.9m)')
ax.plot([x0, x1], [y0, y1], 'o-', color='#f59e0b', lw=3, label=f'Elo 1 (L1=0.5m, θ1={theta1_deg}°)')
ax.plot([x1, x2], [y1, y2], 's-', color='#d97706', lw=3, label=f'Elo 2 (L2=0.4m, θ2={theta2_deg}°)')
ax.plot(x2, y2, 'o', color='#22d3ee', markersize=9, label=f'Efetuador ({x2:.3f}, {y2:.3f})m')
ax.set_xlim(-1, 1); ax.set_ylim(-0.2, 1)
ax.set_title('Cinemática Direta e Workspace do Manipulador 2R', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Eixo X (m)', color='#94a3b8', fontsize=9)
ax.set_ylabel('Eixo Y (m)', color='#94a3b8', fontsize=9)
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#334862', alpha=0.5, ls=':')
ax.legend(facecolor='#243247', edgecolor='#334862', labelcolor='#ffffff', fontsize=7.5)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=200, facecolor=fig.get_facecolor())
plt.close()
```
