> **Enunciado Original:**
> Dimensionar os limites de velocidade e desaceleração de segurança para um robô colaborativo (Cobot) atuando em espaço compartilhado homem-máquina conforme os requisitos da norma técnica ISO/TS 15066 (Power and Force Limiting - PFL). Calcular a distância de parada mínima e a força de contato transitório permissível para a região do antebraço humano com massa equivalente do manipulador m = 25 kg.

### Formulação Matemática Formal
$$S_p = v \cdot T_r + \frac{v^2}{2a} + C, \quad F = v \cdot \sqrt{k \cdot \mu_{\text{red}}}, \quad \mu_{\text{red}} = \frac{m_R \cdot m_H}{m_R + m_H}$$


### Passo 1: Princípios da Norma ISO/TS 15066 (PFL)
A norma estabelece 4 modos de colaboração: Parada monitorada segura, Guia manual, Monitoramento de velocidade/separação e Limitação de potência e força. No modo PFL, contatos físicos são tolerados desde que abaixo dos limiares de dor biomecânica.


### Passo 2: Massa Reduzida do Sistema de Impacto
Para antebraço humano (m_H = 2,2\text{ kg}) e robô com carga útil (m_R = 25\text{ kg}): \mu = \frac{25 \times 2,2}{25 + 2,2} = \frac{55}{27,2} \approx 2,022\text{ kg}.


### Passo 3: Limite de Força de Contato Transitório
O limiar permissível de força para contato quase-estático no braço é F_{max} = 140\text{ N} e rigidez elástica superficial de k = 40\text{ N/mm} = 40.000\text{ N/m}. A velocidade máxima de aproximação é v_{max} = \frac{F_{max}}{\sqrt{k \cdot \mu}} = \frac{140}{\sqrt{40000 \times 2,022}} = \frac{140}{284,4} \approx 0,492\text{ m/s} (492 mm/s).


### Passo 4: Distância de Separação de Segurança (EN ISO 13855)
Considerando tempo de reação do sistema de visão/parada T_r = 0,1\text{ s}, desaceleração de parada a = 2,5\text{ m/s}^2 e margem intrínseca C = 0,1\text{ m}: S = (0,492 \times 0,1) + \frac{0,492^2}{2 \times 2,5} + 0,1 = 0,0492 + 0,0484 + 0,1 = 0,1976\text{ m} (\approx 200 mm).


### Passo 5: Validação e Certificação de Célula Colaborativa
Configuração de scanner laser de área com zona de alerta em 500 mm (redução para 250 mm/s) e zona de parada em 200 mm, assegurando conformidade de nível SIL 2 / PL d.


```python
import numpy as np
import matplotlib.pyplot as plt

v_arr = np.linspace(0.1, 1.2, 200)
# Distancia de parada
tr, a, c = 0.1, 2.5, 0.1
s_dist = v_arr * tr + (v_arr**2) / (2 * a) + c
# Força de contato (antebraço)
mu, k = 2.022, 40000
f_contact = v_arr * np.sqrt(k * mu)
fig, ax1 = plt.subplots(figsize=(6, 3.5), facecolor='#182232')
ax1.set_facecolor('#0f141c')
ax2 = ax1.twinx()
p1 = ax1.plot(v_arr, s_dist * 1000, color='#f59e0b', lw=2.5, label='Distância de Parada (mm)')
p2 = ax2.plot(v_arr, f_contact, color='#ef4444', lw=2.5, label='Força de Impacto (N)')
ax2.axhline(140, color='#06b6d4', ls='--', lw=1.5, label='Limite ISO 15066 (140 N)')
ax1.axvline(0.492, color='#22d3ee', ls=':', lw=1.5, label='V_max Permissível (0.49 m/s)')
ax1.set_xlabel('Velocidade do Efetuador (m/s)', color='#94a3b8', fontsize=9)
ax1.set_ylabel('Distância de Parada (mm)', color='#f59e0b', fontsize=9)
ax2.set_ylabel('Força de Contato (N)', color='#ef4444', fontsize=9)
ax1.tick_params(colors='#94a3b8'); ax2.tick_params(colors='#94a3b8')
ax1.grid(True, color='#334862', alpha=0.5, ls=':')
lines = p1 + p2
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, facecolor='#243247', edgecolor='#334862', labelcolor='#ffffff', fontsize=7.5, loc='upper left')
plt.title('Segurança Robótica Colaborativa (ISO/TS 15066)', color='#ffffff', fontsize=10.5, fontweight='bold')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=200, facecolor=fig.get_facecolor())
plt.close()
```
