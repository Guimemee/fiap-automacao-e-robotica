> **Enunciado Original:**
> Dimensionar o circuito de força e comando para partida de motor de indução trifásico (MIT) de 15 CV, 380 V, 60 Hz, rendimento η = 89% e fp = 0,85. Comparar as correntes de pico e conjugado de partida entre chave de partida direta e partida estrela-triângulo (Y-Δ).

### Formulação Matemática Formal
$$I_n = \frac{P}{\sqrt{3} \cdot V_L \cdot \eta \cdot \cos\varphi}, \quad I_{p(Y)} = \frac{1}{3} I_{p(\Delta)}$$


### Passo 1: Conversão de Potência e Cálculo da Corrente Nominal
Convertendo 15 CV para Watts: P = 15 \times 735,5 = 11.032,5 \text{ W}. Aplicando a fórmula trifásica: I_n = \frac{11.032,5}{\sqrt{3} \times 380 \times 0,89 \times 0,85} \approx 22,1 \text{ A}.


### Passo 2: Análise da Partida Direta (Plena Carga)
Considerando relação típica de corrente de partida I_p/I_n = 7,2: I_{p(\Delta)} = 7,2 \times 22,1 = 159,12 \text{ A}. Esse pico elevado pode causar queda de tensão severa no barramento industrial.


### Passo 3: Dimensionamento da Partida Estrela-Triângulo (Y-Δ)
Na configuração estrela (Y), a tensão de fase é V_F = V_L / \sqrt{3}. A corrente de fase reduz para 1/\sqrt{3} e a corrente de linha reduz por fator de 3: I_{p(Y)} = \frac{159,12}{3} \approx 53,04 \text{ A}.


### Passo 4: Conjugado Mecânico de Partida
Como o torque é proporcional ao quadrado da tensão aplicada (C \propto V^2), temos C_{(Y)} = \frac{1}{3} C_{(\Delta)}. Portanto, a partida deve ocorrer sem carga mecânica nominal acoplada.


### Passo 5: Especificação dos Contatores e Relé de Sobrecarga
Contator K1 e K2 dimensionados para I_c = 0,58 \times I_n = 12,82 \text{ A}. Contator de estrela K3 para 0,33 \times I_n = 7,3 \text{ A}. Relé térmico ajustado na faixa de 18-25 A.


```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 5, 500)
# Curva de corrente: partida direta vs estrela-triangulo
i_dir = 159.12 * np.exp(-t / 0.8) + 22.1 * (1 - np.exp(-t / 0.8))
i_yd = np.where(t < 2.5, 53.04 * np.exp(-t / 0.8) + 7.4 * (1 - np.exp(-t / 0.8)), 
               np.maximum(22.1, 45.0 * np.exp(-(t - 2.5) / 0.4) + 22.1))
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#182232')
ax.set_facecolor('#0f141c')
ax.plot(t, i_dir, label='Partida Direta (Pico = 159.1 A)', color='#ef4444', lw=2)
ax.plot(t, i_yd, label='Estrela-Triângulo (Pico = 53.0 A)', color='#f59e0b', lw=2)
ax.axvline(2.5, color='#06b6d4', ls='--', alpha=0.7, label='Transição Y -> Δ (t=2.5s)')
ax.set_title('Perfil Transitório de Corrente na Partida de Motor de 15 CV', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Tempo (segundos)', color='#94a3b8', fontsize=9)
ax.set_ylabel('Corrente Eficaz de Linha (A)', color='#94a3b8', fontsize=9)
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#334862', alpha=0.5, ls=':')
ax.legend(facecolor='#243247', edgecolor='#334862', labelcolor='#ffffff', fontsize=8)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=200, facecolor=fig.get_facecolor())
plt.close()
```
