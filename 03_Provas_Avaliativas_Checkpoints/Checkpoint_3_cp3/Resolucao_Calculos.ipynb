> **Enunciado Original:**
> Analisar o comportamento do conjugado e da tensão em regime de controle escalar V/f constante para um inversor de frequência alimentando um motor trifásico de 4 polos em frequências de operação f = [15, 30, 45, 60, 75] Hz. Calcular as velocidades síncronas correspondentes e verificar o efeito do enfraquecimento de campo acima de 60 Hz.

### Formulação Matemática Formal
$$N_s = \frac{120 \cdot f}{P}, \quad \frac{V_1}{f_1} = \frac{V_n}{f_n} = \text{cte} \implies C_m \approx \text{constante}$$


### Passo 1: Princípio da Relação V/f Constante
Para manter o fluxo magnético constante no entreferro (\Phi \propto V/f), a tensão eficaz entregue pelo inversor deve variar linearmente com a frequência de modulação PWM até a frequência base de 60 Hz.


### Passo 2: Cálculo das Velocidades Síncronas
Para motor de 4 polos (P = 4): N_s = \frac{120 \times f}{4} = 30 \times f. Para f = [15, 30, 45, 60, 75] Hz: N_s = [450, 900, 1350, 1800, 2250] RPM.


### Passo 3: Determinação das Tensões de Linha
Com V_n = 380\text{ V} e f_n = 60\text{ Hz}, temos a razão k = 380/60 = 6,33\text{ V/Hz}. Para 15 Hz \rightarrow 95\text{ V}; 30 Hz \rightarrow 190\text{ V}; 45 Hz \rightarrow 285\text{ V}; 60 Hz \rightarrow 380\text{ V}.


### Passo 4: Região de Enfraquecimento de Campo (Field Weakening)
Acima de 60 Hz, a tensão atinge o teto do barramento DC (V_{max} = 380\text{ V}). Para f = 75\text{ Hz}, V permanece 380 V. A relação V/f cai para 5,06 V/Hz e o fluxo magnético se reduz por \Phi \propto 1/f.


### Passo 5: Análise de Conjugado e Potência Mecânica
Na região sub-nominal (0 a 60 Hz), o conjugado nominal é constante (torque constante). Acima de 60 Hz, o conjugado decai com 1/f e o torque máximo com 1/f^2, mantendo a potência nominal constante (potência constante).


```python
import numpy as np
import matplotlib.pyplot as plt

f = np.linspace(5, 90, 200)
v = np.where(f <= 60, 380 * (f / 60), 380)
torque_ratio = np.where(f <= 60, 1.0, 60 / f)
fig, ax1 = plt.subplots(figsize=(6, 3.5), facecolor='#182232')
ax1.set_facecolor('#0f141c')
ax2 = ax1.twinx()
p1 = ax1.plot(f, v, color='#f59e0b', lw=2.5, label='Tensão V (V/f Constante)')
p2 = ax2.plot(f, torque_ratio, color='#06b6d4', lw=2.5, label='Conjugado Relativo (C/Cn)')
ax1.axvline(60, color='#94a3b8', ls='--', alpha=0.7, label='Freq. Base (60 Hz)')
ax1.set_xlabel('Frequência de Alimentação (Hz)', color='#94a3b8', fontsize=9)
ax1.set_ylabel('Tensão de Linha (V)', color='#f59e0b', fontsize=9)
ax2.set_ylabel('Conjugado Normalizado', color='#06b6d4', fontsize=9)
ax1.tick_params(colors='#94a3b8'); ax2.tick_params(colors='#94a3b8')
ax1.grid(True, color='#334862', alpha=0.5, ls=':')
lines = p1 + p2
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, facecolor='#243247', edgecolor='#334862', labelcolor='#ffffff', fontsize=8, loc='center right')
plt.title('Característica V/f e Curva de Conjugado em Inversor Escalar', color='#ffffff', fontsize=10.5, fontweight='bold')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=200, facecolor=fig.get_facecolor())
plt.close()
```
