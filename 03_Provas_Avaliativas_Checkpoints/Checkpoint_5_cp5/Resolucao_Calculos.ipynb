**Título:** Planejamento de Trajetórias no Espaço das Juntas e Cartesiano

> **Enunciado Original:**
> Projetar uma trajetória suave de junta para um manipulador robótico utilizando polinômio de interpolação cúbica q(t) = a0 + a1*t + a2*t^2 + a3*t^3. A junta deve mover-se de q0 = 10° para qf = 70° em tempo total tf = 3,0 segundos, partindo e terminando em repouso (velocidades inicial e final nulas). Determinar a velocidade e aceleração máximas.

### Formulação Matemática Formal
$$q(t) = a_0 + a_1 t + a_2 t^2 + a_3 t^3, \quad a_2 = \frac{3(q_f - q_0)}{t_f^2}, \quad a_3 = -\frac{2(q_f - q_0)}{t_f^3}$$


### Passo 1: Condições de Contorno de Posição e Velocidade
Temos q(0) = q_0 = 10^\circ, q(t_f) = q_f = 70^\circ, \dot{q}(0) = 0 e \dot{q}(t_f) = 0. Tempo total t_f = 3\text{ s}.


### Passo 2: Determinação dos Coeficientes Polinomiais
De q(0) = a_0 \implies a_0 = 10. De \dot{q}(0) = a_1 \implies a_1 = 0. Das condições finais: a_2 = \frac{3(70 - 10)}{3^2} = \frac{180}{9} = 20^\circ/\text{s}^2. E a_3 = \frac{-2(60)}{27} = -\frac{40}{9} \approx -4,444^\circ/\text{s}^3.


### Passo 3: Equação Completa da Trajetória de Junta
Posição: q(t) = 10 + 20 t^2 - 4,444 t^3. Velocidade: \dot{q}(t) = 40 t - 13,333 t^2. Aceleração: \ddot{q}(t) = 40 - 26,667 t.


### Passo 4: Cálculo da Velocidade Máxima
A velocidade máxima ocorre quando \ddot{q}(t) = 0 \implies t = 40 / 26,667 = 1,5\text{ s} (exatamente na metade da trajetória). \dot{q}_{max} = 40(1,5) - 13,333(2,25) = 60 - 30 = 30^\circ/\text{s}.


### Passo 5: Análise de Aceleração e Continuidade
Aceleração máxima ocorre no instante inicial t=0 (\ddot{q}_{max} = +40^\circ/\text{s}^2) e desaceleração máxima em t=3 s (\ddot{q}_{min} = -40^\circ/\text{s}^2). Trajetória contínua em C^1 e C^2 garantindo suavidade mecânica.


```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 3, 300)
q0, qf, tf = 10, 70, 3
a2 = 3 * (qf - q0) / (tf**2)
a3 = -2 * (qf - q0) / (tf**3)
pos = q0 + a2 * t**2 + a3 * t**3
vel = 2 * a2 * t + 3 * a3 * t**2
acc = 2 * a2 + 6 * a3 * t
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(6, 4.2), facecolor='#182232', sharex=True)
for ax in (ax1, ax2, ax3):
    ax.set_facecolor('#0f141c')
    ax.tick_params(colors='#94a3b8')
    ax.grid(True, color='#334862', alpha=0.5, ls=':')
ax1.plot(t, pos, color='#f59e0b', lw=2)
ax1.set_ylabel('Posição (°)', color='#f59e0b', fontsize=8.5)
ax1.set_title('Perfil de Trajetória Polinomial Cúbica de Junta', color='#ffffff', fontsize=10, fontweight='bold')
ax2.plot(t, vel, color='#06b6d4', lw=2)
ax2.set_ylabel('Velocidade (°/s)', color='#06b6d4', fontsize=8.5)
ax3.plot(t, acc, color='#ec4899', lw=2)
ax3.set_ylabel('Aceleração (°/s²)', color='#ec4899', fontsize=8.5)
ax3.set_xlabel('Tempo (segundos)', color='#94a3b8', fontsize=8.5)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=200, facecolor=fig.get_facecolor())
plt.close()
```
