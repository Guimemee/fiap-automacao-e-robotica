> **Enunciado Original:**
> Realizar o estudo de viabilidade técnico-financeira para implantação de uma célula robótica de soldagem/paletização substituindo uma operação manual com 3 operadores por turno em 2 turnos diários. O investimento inicial do robô é CAPEX = R$ 450.000,00 com custo de manutenção/energia OPEX = R$ 3.500,00/mês. O custo de cada operador é R$ 6.200,00/mês. Determinar o Payback Simples e o VPL para 5 anos a taxa TMA = 12% a.a.

### Formulação Matemática Formal
$$\text{Payback} = \frac{\text{CAPEX}}{\Delta\text{OPEX}_{\text{mensal}}}, \quad \text{VPL} = -I_0 + \sum_{t=1}^{n} \frac{\text{Fluxo}_t}{(1 + \text{TMA})^t}$$


### Passo 1: Levantamento de Custos Operacionais Manuais Atuais
Operação com 6 operadores no total (3 por turno \times 2 turnos): Custo mensal mão de obra = 6 \times R\$ 6.200,00 = R\$ 37.200,00/mês. Custo anual = 12 \times 37.200 = R\$ 446.400,00/ano.


### Passo 2: Custos Operacionais da Célula Robótica
OPEX robô = R\$ 3.500,00/mês. Custo anual robô = 12 \times 3.500 = R\$ 42.000,00/ano. Economia bruta mensal = 37.200 - 3.500 = R\$ 33.700,00/mês (\Delta\text{OPEX} anual = R\$ 404.400,00/ano).


### Passo 3: Cálculo do Payback Simples
Payback = \frac{\text{CAPEX}}{\Delta\text{OPEX}_{\text{mensal}}} = \frac{450.000}{33.700} \approx 13,35 \text{ meses} (\approx 1,11 \text{ anos}).


### Passo 4: Cálculo do Valor Presente Líquido (VPL) a 12% a.a.
Fator de anuidade para n=5 e i=12%: a_{5\rceil 12\%} = \frac{1 - (1,12)^{-5}}{0,12} \approx 3,6048. VPL = -450.000 + (404.400 \times 3,6048) = -450.000 + 1.457.781 = R\$ 1.007.781,00.


### Passo 5: Conclusão de Engenharia e Ganho de OEE
O projeto apresenta excelente atratividade financeira com retorno do investimento em menos de 14 meses e ganho adicional estimado de 18% em disponibilidade e qualidade da produção (OEE).


```python
import numpy as np
import matplotlib.pyplot as plt

meses = np.arange(0, 37)
capex = 450000
custo_manual_acum = 37200 * meses
custo_robo_acum = capex + 3500 * meses
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#182232')
ax.set_facecolor('#0f141c')
ax.plot(meses, custo_manual_acum / 1000, label='Custo Manual Acumulado (6 Op.)', color='#ef4444', lw=2)
ax.plot(meses, custo_robo_acum / 1000, label='Célula Robótica (CAPEX + OPEX)', color='#06b6d4', lw=2)
# Ponto de payback
payback_m = capex / (37200 - 3500)
ax.axvline(payback_m, color='#f59e0b', ls='--', lw=1.5, label=f'Payback: {payback_m:.1f} meses')
ax.scatter([payback_m], [(capex + 3500*payback_m)/1000], color='#f59e0b', s=60, zorder=5)
ax.set_title('Estudo Econômico de Viabilidade: Homem vs. Robô', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Período de Operação (Meses)', color='#94a3b8', fontsize=9)
ax.set_ylabel('Custo Total Acumulado (R$ Mil)', color='#94a3b8', fontsize=9)
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#334862', alpha=0.5, ls=':')
ax.legend(facecolor='#243247', edgecolor='#334862', labelcolor='#ffffff', fontsize=8)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=200, facecolor=fig.get_facecolor())
plt.close()
```
